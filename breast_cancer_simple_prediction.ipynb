# Breast Cancer Prediction - Simple ML Project

This notebook uses the **1,000-patient dummy dataset** provided with this project.

**Goal:** load the CSV → split the data → train a model → evaluate it → test predictions.

## 1. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

## 2. Load the dummy patient dataset

Keep `breast_cancer_dummy_dataset.csv` in the same folder as this notebook.

```python
data = pd.read_csv("breast_cancer_dummy_dataset.csv")

print("Number of patients:", len(data))
print("Number of columns:", len(data.columns))

data.head()
```

## 3. Prepare the data

`actual_label_for_testing` is the target we want the model to predict.

`patient_id` is only an identifier, so we remove it.

```python
X = data.drop(columns=["patient_id", "actual_label_for_testing"])
y = data["actual_label_for_testing"]

print("Features:", X.shape[1])
print("Target classes:")
print(y.value_counts())
```

## 4. Split into training and testing data

80% of the patients are used for training and 20% are kept aside for testing.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training patients:", len(X_train))
print("Testing patients:", len(X_test))
```

## 5. Build the machine learning model

We use a **Random Forest Classifier**. It is simple to use and works well for many tabular classification problems.

```python
model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

model.fit(X_train, y_train)

print("Model training completed!")
```

## 6. Evaluate the model

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

## 7. Confusion matrix

```python
cm = confusion_matrix(y_test, y_pred, labels=["Malignant", "Benign"])

print("Confusion Matrix:")
print(cm)

plt.imshow(cm)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks([0, 1], ["Malignant", "Benign"])
plt.yticks([0, 1], ["Malignant", "Benign"])

for i in range(2):
    for j in range(2):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.show()
```

## 8. Test one patient

Here we take one patient from the test set and ask the trained model for a prediction.

```python
patient = X_test.iloc[[0]]
actual_result = y_test.iloc[0]
predicted_result = model.predict(patient)[0]

print("Actual result:", actual_result)
print("Predicted result:", predicted_result)
```

## 9. Test a completely new patient

You can enter the 30 feature values of a new patient here. For an easy test, this example uses the first row of the dummy CSV.

```python
new_patient = data.drop(columns=["patient_id", "actual_label_for_testing"]).iloc[[0]]

prediction = model.predict(new_patient)[0]
probability = model.predict_proba(new_patient).max()

print("Prediction:", prediction)
print("Model confidence:", round(probability * 100, 2), "%")
```

## 10. Save the trained model

This lets you use the trained model later without training it again.

```python
import joblib

joblib.dump(model, "breast_cancer_model.pkl")

print("Model saved as breast_cancer_model.pkl")
```
